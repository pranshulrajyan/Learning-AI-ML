```python
# What is the result of the following expression?
# 0 * np.nan
# np.nan == np.nan
# np.inf > np.nan
# np.nan - np.nan
# np.nan in set([np.nan])
# 0.3 == 3 * 0.1
```

```python
#Answers:
#1. nan(Any arithmetic operation involving nan propagates nan)
#2. False(NaN is never equal to anything, including itself)
#3. False(Any comparison (>, <, >=, <=, ==) with nan returns False. nan is unordered — it doesn't fit anywhere on the number line, so it can't be "greater than" or "less than" anything either.)
#4. nan
#5. True(in for a set/dict uses hashing + identity check first, not ==. Since it's the same nan object being checked against itself, Python short-circuits via is before ever hitting __eq__. If you'd instead written np.nan in [float('nan')] (a different nan object, list not set), you'd get False, because list membership uses == linearly, and nan == nan is False.)
#6. False(never compare floats with ==)
```
