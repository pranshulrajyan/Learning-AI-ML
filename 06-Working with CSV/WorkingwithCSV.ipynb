# 4.Index_col parameter

```python
df = pd.read_csv('Mall_Customers.csv')
```

```python
df
```

here, the default index and the customerID are both doing the same job as index, so we can remove one of the indexes using the index_col parameter

```python
df = pd.read_csv('Mall_Customers.csv', index_col  = 'CustomerID')
```

```python
df
```

now, we only have one index instead of two.

# 5. usecols parameter

suppose, we only want to use the columns age, customerId and spending score column instead of all the columns in the data, so we use the usecols parameter to do so.

```python
df = pd.read_csv('Mall_Customers.csv', usecols = ['CustomerID','Age','Spending Score (1-100)'], index_col = 'CustomerID')
```

```python
df
```
